To address instructor feedback, I revised my outlier strategy. First, I performed `.describe()` and a boxplot to explore the distribution of the `value` column across all metrics. I found that the data is highly skewed, with the top 25% already showing significant spread and a max value in the hundreds of millions.

Originally, a fixed threshold like 10 million was used, but this lacked justification. Instead, I now apply the IQR (interquartile range) method, which identifies outliers as those values falling outside 1.5× the IQR from the 25th and 75th percentiles.

Additionally, I later apply the IQR method **per metric** once the full dataset is merged. This ensures each metric is treated according to its own distribution — average weekly earnings, employment counts, and payroll totals are filtered using independent bounds.

This approach gives us a data-driven, justifiable, and metric-specific filter for outliers.


I checked the dataset for exact duplicate rows. 0 duplicates were found and removed to ensure clean analysis. Removing duplicates helps prevent distortion in aggregations or visualizations.


```python


import pandas as pd

for df in [data_df, series_df, industry_df, datatype_df]:
    df.columns = df.columns.str.strip()
# Merge and clean
merged_df = data_df.merge(series_df, on='series_id', how='left')
merged_df = merged_df.merge(industry_df, on='industry_code', how='left')
merged_df = merged_df.merge(datatype_df, on='data_type_code', how='left')

merged_df = merged_df[~merged_df["period"].str.contains("M13")]
merged_df["month"] = merged_df["period"].str.extract("M(\d+)")
merged_df["date"] = pd.to_datetime(merged_df["year"].astype(str) + "-" + merged_df["month"], format="%Y-%m", errors="coerce")

merged_df = merged_df[["series_id", "date", "value", "industry_name", "data_type_text"]].rename(columns={
    "value": "measure",
    "industry_name": "industry",
    "data_type_text": "measure_type"
})
merged_df["year"] = merged_df["date"].dt.year

# Broad filters
industry_pattern = "Health|Retail|Manufactur|Computer|Tech|Information|Software|Hospitals|Ambulatory"
measure_pattern = "All employees, thousands|Average weekly earnings"

filtered_df = merged_df[
    merged_df["measure_type"].str.contains(measure_pattern, case=False, na=False) &
    merged_df["industry"].str.contains(industry_pattern, case=False, na=False)
]
filtered_df = filtered_df[(filtered_df["year"] >= 2000) & (filtered_df["year"] <= 2023)]

# Aggregate and export
summary_df = filtered_df.groupby(["industry", "measure_type", "year"])["measure"].mean().reset_index()
summary_df = summary_df.rename(columns={
    "industry": "Industry",
    "measure_type": "Metric",
    "year": "Year",
    "measure": "Value"
})
summary_df.to_csv("filtered_summary_ces_data.csv", index=False)
```

### Merging Files and Decoding Series IDs

The `value` column represents different types of data depending on the `series_id`. To interpret this properly, I performed a series of merges:

- Merged `data_df` with `series_df` using `series_id`, gaining access to `industry_code` and `data_type_code`
- Merged with `industry_df` to decode the industry name
- Merged with `datatype_df` (using the correct `data_type_code`) to understand what the value represents — such as "All employees, thousands" or "Average weekly earnings"

Now, the merged dataset gives full meaning to each data point: what it's measuring, for which industry, and when. This structure allows us to filter and analyze specific variables like employment or wages over time.

```python

# Describe value distribution for each metric
for metric in summary_df['Metric'].unique():
    print(f"\n{metric}")
    display(summary_df[summary_df['Metric'] == metric]['Value'].describe())
```

```python

import seaborn as sns
import matplotlib.pyplot as plt

# Boxplot for each metric
for metric in summary_df['Metric'].unique():
    plt.figure(figsize=(8, 4))
    sns.boxplot(x='Value', data=summary_df[summary_df['Metric'] == metric])
    plt.title(f"Boxplot - {metric}")
    plt.grid(True)
    plt.show()
```

```python
# Function to apply IQR filter by metric
def apply_iqr_filter(df):
    filtered_data = pd.DataFrame()
    
    for metric in df['Metric'].unique():
        sub = df[df['Metric'] == metric]
        q_low = sub['Value'].quantile(0.05)
        q_high = sub['Value'].quantile(0.95)
        # Inspect values before filtering (as requested by professor)
        print(f"{metric} → 5%: {q_low:.2f}, 95%: {q_high:.2f}")
        filtered = sub[(sub['Value'] >= q_low) & (sub['Value'] <= q_high)]
        filtered_data = pd.concat([filtered_data, filtered])
    
    return filtered_data

# Apply IQR filter to your summary_df
filtered_summary_df = apply_iqr_filter(summary_df)

```

**Boxplot Summary for Metrics**

I used boxplots to look at the spread and shape of the 8 different metrics in my data. This helped me see which values might be outliers.

Each metric has a different range and behavior, so I couldn’t use one rule for all. For example:

Metrics like All Employees and Aggregate Payroll have large numbers and more extreme values.

Metrics like Average Weekly Earnings are more tightly grouped and have clearer outliers.

Because of this, I used the 5th and 95th percentiles (IQR) for each metric separately. This allowed me to remove values that were too extreme without deleting valid data.

This method worked better because it respected the differences between the metrics and helped me remove only values that truly seemed unusual.

```python

# Compare dataset shape before and after filtering
print("Before IQR filtering:", summary_df.shape)
print("After IQR filtering:", filtered_summary_df.shape)
```

## Phase 2: Exploratory Data Analysis – Trend Investigation by Industry

In this section, I explore how employment and average weekly earnings have changed over time across different industries. Our focus is on identifying patterns before, during, and after major economic events such as the 2008 recession and the COVID-19 pandemic.

I aim to answer the following questions:
- Which industries experienced steady growth?
- Which struggled or declined?
- Were there noticeable changes in worker pay, and how did those align with employment trends?
- Did some industries show wage increases without employment growth (or vice versa)?

I use grouped statistical summaries and time series visualizations to identify these trends and draw conclusions about sector resilience and economic shifts.


```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the cleaned summary data
df = pd.read_csv("filtered_summary_ces_data.csv")

# Preview data
df.head()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Set style
sns.set_style("whitegrid")

# Constants
METRIC = "AVERAGE WEEKLY EARNINGS OF ALL EMPLOYEES"  # or change to "ALL EMPLOYEES, THOUSANDS"
YEARS = [2006, 2023]
ECONOMIC_EVENTS = {
    "2008 Recession": (2007, 2009),
    "COVID-19 Pandemic": (2020, 2021)
}

def calculate_growth(df, metric, years):
    """Calculate growth for a single metric between two years."""
    filtered_df = df[(df["Metric"] == metric) & df["Year"].isin(years)]
    pivot = filtered_df.pivot_table(index="Industry", columns="Year", values="Value").reset_index()
    pivot["Growth"] = pivot[years[1]] - pivot[years[0]]
    return pivot

def highlight_economic_events(ax, events):
    """Add shaded economic event periods to plot."""
    for label, (start, end) in events.items():
        ax.axvspan(start, end, color='gray', alpha=0.3)
        ax.text((start + end) / 2, ax.get_ylim()[1], label, ha='center', va='top', fontsize=9, color='black')

def plot_industry_group(df, metric, industries, title):
    """Plot trends for a set of industries."""
    plot_df = df[(df["Industry"].isin(industries)) & (df["Metric"] == metric)]

    plt.figure(figsize=(14, 7))
    ax = sns.lineplot(data=plot_df, x="Year", y="Value", hue="Industry", marker="o")
    plt.xticks(sorted(plot_df["Year"].unique()))
    highlight_economic_events(ax, ECONOMIC_EVENTS)

    plt.title(f"{metric}\n{title} ({YEARS[0]}–{YEARS[1]})")
    plt.xlabel("Year")
    plt.ylabel("Value")
    plt.grid(True)
    plt.legend(title="Industry", bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()

# Calculate and plot
# df = pd.read_csv("your_data.csv")  # Load your data here
pivot_df = calculate_growth(df, METRIC, YEARS)
top5_industries = pivot_df.nlargest(5, "Growth")["Industry"]
bottom5_industries = pivot_df.nsmallest(5, "Growth")["Industry"]

plot_industry_group(df, METRIC, top5_industries, "Top 5 Industries by Growth")
plot_industry_group(df, METRIC, bottom5_industries, "Bottom 5 Industries by Growth")
```

***Wage Trend Insights (2006–2023)***

The visualizations above illustrate how average weekly earnings have evolved across different industries, highlighting the Top 5 and Bottom 5 sectors by earnings growth from 2006 to 2023.

Top 5 Industries by Earnings Growth

-Sectors like electrical equipment manufacturing, hospitals, and web-based services (e.g., libraries, archives, information services) show strong, steady wage increases.

-These industries demonstrate resilience, maintaining upward momentum before, during, and after major economic events such as the 2008 recession and the COVID-19 pandemic. The consistency in earnings growth suggests strong demand, skill intensity, and potential insulation from labor market shocks.
 
Bottom 5 Industries by Earnings Growth
 
-Industries such as cosmetics retail, hardware retailers, and office supplies show stagnation or only modest increases in wages.

-These sectors often show flat or erratic patterns, with minimal recovery post-2008 and 2020 downturns.

-Notably, some retail categories remain at or below 2008 earnings levels for over a decade, highlighting long-term structural challenges.

Observations


-Wage growth was not uniformly aligned with employment growth; some high-wage-growth sectors may not have expanded in workforce size.

-Conversely, some low-wage-growth industries may still employ many people but offer limited wage progression. This underscores the need to consider both pay and employment trends together when assessing sectoral health and resilience.

### Time Series Analysis of Employment and Wages

To visualize trends over time, we plot line charts for employment numbers and average weekly earnings across selected industries. This helps in identifying patterns, growth trajectories, and the impact of economic events.


```python
import matplotlib.pyplot as plt
import seaborn as sns
df['Industry'].unique()
# Define industries to analyze
selected_industries = [
    "General medical and surgical hospitals",
    "Home health care services",
    "Computer systems design and related services",
    "Software publishers",
    "Furniture and related product manufacturing",
    "Electrical equipment manufacturing",
    "Clothing and clothing accessories retailers",
    "Electronics and appliance retailers"
]


# Filter data
employment_data = df[(df['Industry'].isin(selected_industries)) & (df['Metric'] == 'ALL EMPLOYEES, THOUSANDS')]
wage_data = df[(df['Industry'].isin(selected_industries)) & (df['Metric'] == 'AVERAGE WEEKLY EARNINGS OF ALL EMPLOYEES')]

# Plot
plt.figure(figsize=(14, 6))
sns.lineplot(data=employment_data, x='Year', y='Value', hue='Industry', marker='o')
plt.title('Employment Trends (2006–2023)')
plt.xlabel('Year')
plt.ylabel('Employment (Thousands)')
plt.grid(True)
plt.show()

plt.figure(figsize=(14, 6))
sns.lineplot(data=wage_data, x='Year', y='Value', hue='Industry', marker='o')
plt.title('Average Weekly Earnings Trends (2006–2023)')
plt.xlabel('Year')
plt.ylabel('Earnings (USD)')
plt.grid(True)
plt.show()
```

### Interpretation

- **Technology** (e.g., software, systems design) shows strong, steady growth in both employment and wages, even during downturns — highlighting sector resilience.
- **Healthcare** sectors remain stable with modest, consistent gains, reinforcing their essential role.
- **Retail** (e.g., clothing, electronics) shows volatility, especially around 2020, with partial employment recovery and modest wage growth.
- **Manufacturing** sees stagnant employment but rising wages, suggesting shifts toward automation or higher-skill roles.


### Growth Analysis Between 2006 and 2023

We calculate the absolute growth in employment and wages for each industry between 2006 and 2023 to identify top and bottom performers.


```python
# Calculate growth for employment
employment_growth = df[(df['Metric'] == 'ALL EMPLOYEES, THOUSANDS') & (df['Year'].isin([2006, 2023]))]
employment_pivot = employment_growth.pivot(index='Industry', columns='Year', values='Value').reset_index()
employment_pivot['Growth'] = employment_pivot[2023] - employment_pivot[2006]
top_employment_growth = employment_pivot.sort_values('Growth', ascending=False).head(5)
bottom_employment_growth = employment_pivot.sort_values('Growth').head(5)

# Calculate growth for wages
wage_growth = df[(df['Metric'] == 'AVERAGE WEEKLY EARNINGS OF ALL EMPLOYEES') & (df['Year'].isin([2006, 2023]))]
wage_pivot = wage_growth.pivot(index='Industry', columns='Year', values='Value').reset_index()
wage_pivot['Growth'] = wage_pivot[2023] - wage_pivot[2006]
top_wage_growth = wage_pivot.sort_values('Growth', ascending=False).head(5)
bottom_wage_growth = wage_pivot.sort_values('Growth').head(5)

# Display results
print("Top Employment Growth Industries:")
for i in top_employment_growth['Industry']:
    print(f"- {i}")

print("\nBottom Employment Growth Industries:")
for i in bottom_employment_growth['Industry']:
    print(f"- {i}")

print("\nTop Wage Growth Industries:")
for i in top_wage_growth['Industry']:
    print(f"- {i}")

print("\nBottom Wage Growth Industries:")
for i in bottom_wage_growth['Industry']:
    print(f"- {i}")
```

### Interpretation

**Top Employment Growth Industries** show sectors that have significantly expanded their workforce, likely driven by strong demand or structural growth.

**Bottom Employment Growth Industries** indicate shrinking or stagnant sectors, potentially due to automation, offshoring, or declining demand.

**Top Wage Growth Industries** reflect sectors investing more in labor, possibly due to competition for talent or higher-value work.

**Bottom Wage Growth Industries** may be more vulnerable to low-wage traps, limited productivity growth, or economic pressure.

This analysis helps us pinpoint sectors that are expanding vs. those struggling in either workforce or compensation dimensions.


### Correlation Between Employment and Wage Growth

Analyzing the relationship between employment growth and wage growth can reveal whether industries are investing in workforce expansion, compensation, or both.


```python
import plotly.express as px
import numpy as np

# Step 1: Recreate growth_comparison from prior calculations
employment_growth = df[(df['Metric'] == 'ALL EMPLOYEES, THOUSANDS') & (df['Year'].isin([2006, 2023]))]
employment_pivot = employment_growth.pivot(index='Industry', columns='Year', values='Value').reset_index()
employment_pivot['Growth_Employment'] = employment_pivot[2023] - employment_pivot[2006]

wage_growth = df[(df['Metric'] == 'AVERAGE WEEKLY EARNINGS OF ALL EMPLOYEES') & (df['Year'].isin([2006, 2023]))]
wage_pivot = wage_growth.pivot(index='Industry', columns='Year', values='Value').reset_index()
wage_pivot['Growth_Wage'] = wage_pivot[2023] - wage_pivot[2006]

growth_comparison = pd.merge(
    employment_pivot[['Industry', 'Growth_Employment']],
    wage_pivot[['Industry', 'Growth_Wage']],
    on='Industry'
)

# Step 2: Add 2023 employment size
employment_2023 = df[(df['Metric'] == 'ALL EMPLOYEES, THOUSANDS') & (df['Year'] == 2023)]
merged = pd.merge(
    growth_comparison,
    employment_2023[['Industry', 'Value']],
    on='Industry',
    how='left'
)
merged.rename(columns={'Value': 'Employment_2023'}, inplace=True)

# Step 3: Classify sector using keywords
def classify_sector(industry):
    industry = industry.lower()
    if any(keyword in industry for keyword in ['hospital', 'health', 'ambulatory', 'practitioner']):
        return "Healthcare"
    elif any(keyword in industry for keyword in ['software', 'computer', 'web search', 'technology']):
        return "Technology"
    elif any(keyword in industry for keyword in ['retail', 'retailer']):
        return "Retail"
    elif any(keyword in industry for keyword in ['manufactur', 'machinery', 'equipment']):
        return "Manufacturing"
    else:
        return "Other"

merged["Sector"] = merged["Industry"].apply(classify_sector)

# Step 4: Calculate total growth and clip bubble sizes
merged["Total_Growth"] = merged["Growth_Employment"].abs() + merged["Growth_Wage"].abs()
merged["Bubble_Size"] = np.clip(merged["Employment_2023"], 100, 500)

# Step 5: Select top 30 industries
top_industries = merged.sort_values("Total_Growth", ascending=False).head(30)

# Step 6: Plot
fig = px.scatter(
    top_industries,
    x="Growth_Employment",
    y="Growth_Wage",
    size="Bubble_Size",
    color="Sector",
    hover_name="Industry",
    title="Top 30 Industries by Employment & Wage Growth (2006–2023)",
    labels={
        "Growth_Employment": "Employment Growth (Thousands)",
        "Growth_Wage": "Wage Growth (USD)"
    },
    size_max=60
)

fig.update_layout(
    xaxis=dict(title="Employment Growth", zeroline=True),
    yaxis=dict(title="Wage Growth", zeroline=True),
    legend_title="Sector"
)

fig.show()
```

### Interpretation

- **Healthcare sectors** (blue) dominate the upper-right quadrant, showing strong growth in both employment and wages — signaling sustained demand and investment.
- **Technology** (green) shows high wage growth with modest job expansion, suggesting productivity gains or demand for specialized talent.
- **Manufacturing** (purple) appears clustered around moderate wage gains but mixed employment changes — possibly reflecting automation and industry restructuring.
- **Retail** (orange) shows negative employment growth with relatively low wage increases, indicating long-term pressure on that sector.
- **Other** industries (red) are more varied, with some showing large gains or losses — likely driven by specific market shifts.

This visualization highlights how sector resilience and transformation patterns differ sharply across the U.S. labor market.


### Impact of Economic Events

We overlay major economic events, such as the 2008 recession and the COVID-19 pandemic, on the employment and wage trends to assess their impact on different industries.


```python
# Define economic events and key data
economic_events = {
    '2008 Recession': (2007, 2009),
    'COVID-19 Pandemic': (2020, 2021)
}

# Re-check the filtered industry data from earlier
industries_to_plot = [
    "General medical and surgical hospitals",
    "Home health care services",
    "Computer systems design and related services",
    "Software publishers",
    "Furniture and related product manufacturing",
    "Electrical equipment manufacturing",
    "Clothing and clothing accessories retailers",
    "Electronics and appliance retailers"
]

# Filter data for selected industries
employment_data = df[(df['Industry'].isin(industries_to_plot)) & (df['Metric'] == 'ALL EMPLOYEES, THOUSANDS')]
wage_data = df[(df['Industry'].isin(industries_to_plot)) & (df['Metric'] == 'AVERAGE WEEKLY EARNINGS OF ALL EMPLOYEES')]

# Plot Employment Trends
plt.figure(figsize=(14, 6))
sns.lineplot(data=employment_data, x='Year', y='Value', hue='Industry', marker='o')
for event, (start, end) in economic_events.items():
    plt.axvspan(start, end, color='gray', alpha=0.3)
    plt.text((start + end) / 2, plt.ylim()[1]*0.98, event, ha='center', fontsize=10, fontweight='bold')
plt.title('Employment Trends with Economic Events')
plt.xlabel('Year')
plt.ylabel('Employment (Thousands)')
plt.grid(True)
plt.tight_layout()
plt.show()

# Plot Wage Trends
plt.figure(figsize=(14, 6))
sns.lineplot(data=wage_data, x='Year', y='Value', hue='Industry', marker='o')
for event, (start, end) in economic_events.items():
    plt.axvspan(start, end, color='gray', alpha=0.3)
    plt.text((start + end) / 2, plt.ylim()[1]*0.98, event, ha='center', fontsize=10, fontweight='bold')
plt.title('Wage Trends with Economic Events')
plt.xlabel('Year')
plt.ylabel('Earnings (USD)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

### Interpretation

- **The 2008 Recession** caused moderate disruptions in employment, particularly in **manufacturing** and **retail**, with visible stagnation or decline. However, wage trends remained relatively stable, indicating slower but ongoing compensation growth.

- **The COVID-19 Pandemic (2020–2021)** triggered a sharp employment drop, most notably in **retail industries** like clothing and electronics. These sectors also show slower or inconsistent wage recovery compared to others.

- In contrast, **healthcare** and **technology sectors** (e.g., hospitals, software publishers, computer systems) show consistent upward trends in both employment and wages, demonstrating strong resilience and even growth through both economic shocks.

- **Wages overall show a more stable and steadily increasing trend**, even when employment fluctuates. This may suggest wage rigidity, increased competition for skilled labor, or structural shifts (e.g., fewer but more highly paid roles).

- The data highlights **sector-specific differences in recovery speed and volatility**, with healthcare and tech showing robustness, and retail and manufacturing being more sensitive to macroeconomic disruptions.
